# Bernoulli Distribution.

Whenever there is exactly one trial and only two possible outcomes, we use the Bernoulli Distribution.

---------------------------------------------------------------------------------
Bernoulli needs exactly two mutually exclusive outcomes.

Usually(Success / Failure,Yes / No,Pass / Fail,Good / Bad)

example :

suppose 1000 emails in which 30% (p=0.3) spam.How many spam emails?

ans---> no_of_spams = **300**


what is the average spam ?

mean = (300/1000)----> **0.3**


**mean=p**

-----------------

How much do the outcomes vary?

    Variance=p(1−p)

**example**

X = 1 → Spam email

X = 0 → Not spam

p = 0.3 (30% of emails are spam)

```python
# PMF : Probability Mass Function for discrete distributions,

from scipy.stats import bernoulli

bernoulli.pmf(1,0.3) # What is the probability that X equals 1?
# Probability that an email is spam.
```

```python
bernoulli.pmf(0,0.3) # What is the probability that X equals 0?
# Probability that an email is NOT spam.
```

```python
#CDF : Cumulative Distribution Function.(Cumulative means adding probabilities.)
#      P(X≤1)=P(0)+P(1)=0.7+0.3=1

bernoulli.cdf(1,0.3)# Probability that the email is either NOT spam or spam.
```

```python
# SF : Survival Function. The probability greater than that value.
#   P(X>x)=1−CDF(x)

bernoulli.sf(0,0.3) # Only the value 1 is greater than 0.
# Probability that the email is spam.
```

# Binomial Distribution

A Binomial Distribution is simply many independent Bernoulli trials combined.

n->How many experiments did we perform?

x->How many successes occurred?

Binomial works only if:

1. Fixed number of trials

2. Only two outcomes

3. Independent trials

4. Constant Probability




#### **mean**

1. 10 tosses having 50%(p=0.5) Head.Expected Heads?

ans --> 5

2. 100 tosses having 20%(p=0.2) Head.Expected heads?

ans --> 20

**mean = n*p**

---------------------------------------------------------------------

#### **variance** = **np(1−p)**

```python
# suppose A company sends 10 promotional emails.Historically,30% become spam

from scipy.stats import binom


# Probability that EXACTLY 3 out of 10 emails become spam.
a=binom.pmf(3,10,0.3) #  3 successes out of 10 trials.
print(a)

# Probability that AT MOST 3 out of 10 emails become spam.
b=binom.cdf(3,10,0.3) # At most 3 successes.
print(b)

# Probability that MORE THAN 3 out of 10 emails become spam.
c=binom.sf(3,10,0.3) # Greater than 3
print(c)
```

**Note** : flip a coin once

1. Did you get Head? **Bernoulli**

2. How many Heads did you get? **Binomial** (ounting successes across many independent trials.)

# Poisson Distribution

Poisson Distribution is used to count how many times an event occurs in a fixed interval of time or space.

**λ --> Average number of events in the chosen interval**


Conditions:
 
1. Events occur randomly. ( Calls don't arrive exactly every 2 minutes.The timing is unpredictable.)

2. Events are independent.

3. Constant average rate.

**mean=λ**

**Variance=λ**  (Variation naturally occurs around the average.)

```python
from scipy.stats import poisson

# Probability of receiving EXACTLY 3 calls in one hour,
a=poisson.pmf(3,5) # Exactly 3 calls when average is 5 calls/hour.
print(a)

b=poisson.cdf(3,5) #  Probability of receiving At most 3 calls.

c=poisson.sf(3,5) # Probability of receiving More than 3 calls.
```
