# Bernoulli Distribution.

Whenever there is exactly one trial and only two possible outcomes, we use the Bernoulli Distribution.

---------------------------------------------------------------------------------
Bernoulli needs exactly two mutually exclusive outcomes.

Usually(Success / Failure,Yes / No,Pass / Fail,Good / Bad)

example :

suppose 1000 emails in which 30% (p=0.3) spam.How many spam emails?

ans---> no_of_spams = **300**


what is the average spam ?

mean = (300/1000)----> **0.3**


**mean=p**

-----------------

How much do the outcomes vary?

    Variance=p(1−p)

**example**

X = 1 → Spam email

X = 0 → Not spam

p = 0.3 (30% of emails are spam)

```python
# PMF : Probability Mass Function for discrete distributions,

from scipy.stats import bernoulli

bernoulli.pmf(1,0.3) # What is the probability that X equals 1?
# Probability that an email is spam.
```

```python
bernoulli.pmf(0,0.3) # What is the probability that X equals 0?
# Probability that an email is NOT spam.
```

```python
#CDF : Cumulative Distribution Function.(Cumulative means adding probabilities.)
#      P(X≤1)=P(0)+P(1)=0.7+0.3=1

bernoulli.cdf(1,0.3)# Probability that the email is either NOT spam or spam.
```

```python
# SF : Survival Function. The probability greater than that value.
#   P(X>x)=1−CDF(x)

bernoulli.sf(0,0.3) # Only the value 1 is greater than 0.
# Probability that the email is spam.
```

# Binomial Distribution

A Binomial Distribution is simply many independent Bernoulli trials combined.

n->How many experiments did we perform?

x->How many successes occurred?

Binomial works only if:

1. Fixed number of trials

2. Only two outcomes

3. Independent trials

4. Constant Probability




#### **mean**

1. 10 tosses having 50%(p=0.5) Head.Expected Heads?

ans --> 5

2. 100 tosses having 20%(p=0.2) Head.Expected heads?

ans --> 20

**mean = n*p**

---------------------------------------------------------------------

#### **variance** = **np(1−p)**

```python
# suppose A company sends 10 promotional emails.Historically,30% become spam

from scipy.stats import binom


# Probability that EXACTLY 3 out of 10 emails become spam.
a=binom.pmf(3,10,0.3) #  3 successes out of 10 trials.
print(a)

# Probability that AT MOST 3 out of 10 emails become spam.
b=binom.cdf(3,10,0.3) # At most 3 successes.
print(b)

# Probability that MORE THAN 3 out of 10 emails become spam.
c=binom.sf(3,10,0.3) # Greater than 3
print(c)
```

# Poisson Distribution

Poisson Distribution is used to count how many times an event occurs in a fixed interval of time or space.

**λ --> Average number of events in the chosen interval**


Conditions:
 
1. Events occur randomly. ( Calls don't arrive exactly every 2 minutes.The timing is unpredictable.)

2. Events are independent.

3. Constant average rate.

**mean=λ**

**Variance=λ**  (Variation naturally occurs around the average.)

```python
from scipy.stats import poisson

# Probability of receiving EXACTLY 3 calls in one hour,
a=poisson.pmf(3,5) # Exactly 3 calls when average is 5 calls/hour.
print(a)

b=poisson.cdf(3,5) #  Probability of receiving At most 3 calls.

c=poisson.sf(3,5) # Probability of receiving More than 3 calls.
```

# 2. Continuous Distributions

A continuous probability distribution is used when a random variable can take any value within a range.

types of continuous distributions are:

    Uniform Distribution
    
    Normal Distribution

1. Uniform Distribution

A Uniform Distribution is a continuous probability distribution where **every value within a given interval is equally likely.**  (has the same probability density.)

                                or
                                
(A Uniform Distribution is a distribution where all values within a given range are equally likely.)

Probability = Length of required interval / Length of total interval

Example:

**Random number generator** that generates a number between 0 and 10.(Assuming that every number between 0 and 10 has an equal chance of being selected.)

**generated_numbers :** (0.5,2.3,4.7,7.1,9.8)

```python
# probability density at a particular point not probability.

from scipy.stats import uniform

uniform.pdf(5, loc=0, scale=10)
```

```python
# What is the probability that the generated number is less than or equal to 5?

from scipy.stats import uniform

uniform.cdf(5, loc=0, scale=10)

# There is a 0.5 probability, or 50% chance, that the randomly generated number will be less than or equal to 5.
```

```python
# What is the probability that the randomly generated number is greater than 5?

uniform.sf(5, loc=0, scale=10)
```

Example :
bus will arrive sometime between 10:00 AM and 10:10 AM.

10:00 ────────────────────────── 10:10

                Bus can arrive
                anywhere here


**What is the probability that the bus arrives between 10:02 AM and 10:04 AM?**

The total time range is:10 minutes

The required range is: 2 minutes

P(10:02  ≤  X  ≤ 10:04) = 2 / 10  = 20%

-------------------------------------------------------------------------

The bus can arrive anywhere in a 10-minute interval. 

( the focus is on 2-minute interval ) . Therefore, the probability is 2 divided by 10, which is 20%

```python
from scipy.stats import uniform

# Bus arrival time is uniformly distributed
# between 0 and 10 minutes after 10:00 AM

# Probability that bus arrives between 10:02 and 10:04

uniform.cdf(4, loc=0, scale=10) - uniform.cdf(2, loc=0, scale=10)
# 20% probability that the bus will arrive between 10:02 AM and 10:04 AM.
```

```python
# What is the probability that the bus arrives after 10:04 AM?

from scipy.stats import uniform

uniform.sf(4, loc=0, scale=10)
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import uniform

x = np.linspace(0, 10, 1000)
y = uniform.pdf(x, 0, 10)

sns.lineplot(x=x, y=y)
plt.fill_between(x, y, where=(x >= 2) & (x <= 4))

plt.xticks([0, 2, 4, 10], 
           ["10:00", "10:02", "10:04", "10:10"])

plt.xlabel("Bus Arrival Time")
plt.ylabel("Probability Density")
plt.title("Probability of Bus Arriving Between 10:02 and 10:04")
plt.show()
```

## Normal Distribution

The Normal Distribution is a continuous probability distribution that is:

    Bell-shaped

    Symmetric
    
    Centered around the mean

**The probability density is highest around the mean and decreases as we move away from the mean.**

```python
#PDF:The PDF value represents the height of the bell curve at 60.

# example :  The marks of students are approximately normally distributed.

# Mean = 60 marks
# Standard Deviation = 10 marks

from scipy.stats import norm
norm.pdf(70, loc=60, scale=10)

# The probability density of the normal distribution at 60 marks is approximately 0.0399.
```

```python
# What is the probability that a student scores 60 marks or less?

norm.cdf(70, loc=60, scale=10)
#Approximately 50% of students are expected to score 60 marks or less.
```

```python
# What is the probability that a student scores more than 60 marks?

norm.sf(70, loc=60, scale=10)
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm

x = np.linspace(20, 100, 1000)
y = norm.pdf(x, 60, 10)

sns.lineplot(x=x, y=y)
plt.fill_between(x, y, where=x <= 70, alpha=0.3, label="CDF: 84.13%")
plt.fill_between(x, y, where=x > 70, alpha=0.3, label="SF: 15.87%")

plt.axvline(70, linestyle="--")
plt.xlabel("Student Marks")
plt.ylabel("Probability Density")
plt.legend()
plt.show()
```
