```python
import numpy as np
import matplotlib.pyplot as plt
```

EXAMPLE:

For a fair six-sided die:Possible outcomes are     [1, 2, 3, 4, 5, 6]

The population mean :    μ=3.5	
	​
	( theoretical population mean )


# Generate 50,000 Dice Rolls

```python
np.random.seed(42)

dice_rolls = np.random.randint(1, 7, 50000)

print(dice_rolls[:20])
print("Number of rolls:", len(dice_rolls))
print("Mean:", dice_rolls.mean())
print("Standard deviation:", dice_rolls.std())
```

Theoretical Mean = 3.5

Sample Mean ≈ 3.5

This shows that our 50,000 rolls represent the expected population behavior quite well.

# Visualize the 50,000 Dice Rolls

```python
plt.hist(dice_rolls, bins=np.arange(0.5, 7.5, 1), edgecolor="black")

plt.xlabel("Dice Outcome")
plt.ylabel("Frequency")
plt.title("Distribution of 50,000 Individual Dice Rolls")

plt.show()

# It is approximately uniform because every outcome from 1 to 6 has approximately the same probability.
```

## Taking the Samples and Calculating Sample Means

For a fair die, the population standard deviation is approximately: σ≈1.708 (theoretically)
	​


```python
sample_size = 30
number_of_samples = 5000

sample_means = []

for i in range(number_of_samples):
    sample = np.random.choice(dice_rolls, size=sample_size)
    sample_mean = sample.mean()
    sample_means.append(sample_mean)

sample_means = np.array(sample_means)

print("Number of sample means:", len(sample_means))
print("Mean of sample means:", sample_means.mean())
print("Standard deviation of sample means:", sample_means.std())
```

Population Mean = 3.5

Mean of Sample Means ≈ 3.5

Theoretical Standard Error = 1.708 / √30    ≈ 0.312

Observed Standard Deviation of Sample Means ≈ 0.312

# Sampling Distribution

```python
plt.hist(sample_means, bins=30, edgecolor="black")

plt.axvline(
    sample_means.mean(),
    linestyle="--",
    label="Mean of Sample Means"
)

plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.title("Sampling Distribution of Sample Means (n = 30)")

plt.legend()
plt.show()

# The original dice-roll distribution was approximately uniform, but the distribution of sample means is approximately bell-shaped.This is the Central Limit Theorem.
```

# For Different Sample Sizes

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

dice_rolls = np.random.randint(1, 7, 50000)

sample_sizes = [3, 30, 150]

for n in sample_sizes:

    sample_means = []

    for i in range(5000):
        sample = np.random.choice(dice_rolls, size=n)
        sample_means.append(sample.mean())

    plt.hist(sample_means, bins=30, edgecolor="black")

    plt.axvline(
        np.mean(sample_means),
        linestyle="--"
    )

    plt.title(f"Sampling Distribution of Sample Means (n = {n})")
    plt.xlabel("Sample Mean")
    plt.ylabel("Frequency")

    plt.show()
```

When n = 5

The distribution will be:Relatively wide , Less smooth , More variation in sample means

----------------------------------------------------------------

When n = 30

The distribution will-->Look approximately bell-shaped , Be more concentrated around 3.5

----------------------------------------------------------------
When n = 100

The distribution will:Be much narrower , Be strongly concentrated around 3.5 ,Have a smaller standard error
